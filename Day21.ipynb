```python
def prime(num):
    if num <=1:
        return "Not a prime number"
    
    count=0
    for i in range(1,num+1):
        if num%i==0 :
            count+=1
    if count==2:
        return  num," is prime number" 
    else:
        return "Number is not prime" 
    
num= int(input("enter number for checking:"))
print(prime(num))
```

```python
def large(*args):
    max= args[0]
    for num in args:
        if num > max: 
            max= num
    return max
large(10,20,30,4,8,2,55,67,91,99)        
```

```python
def calculate(*args):
    maximum=max(args)
    minimum=min(args)
    total= sum(args)
    average=total/len(args)

    return maximum, minimum, total, average
calculate(10,20,45,23,5,78,98)
    
```
